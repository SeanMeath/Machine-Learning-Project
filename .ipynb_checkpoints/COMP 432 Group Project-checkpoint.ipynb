Imports:

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
import sklearn
import keras
import extra_keras_datasets
```

Temporary Use of EMNIST Dataset:

```python
from extra_keras_datasets import emnist

train, test = emnist.load_data()
(X_train_em, y_train_em) = train
(X_test_em, y_test_em) = test
```

Dataset Creation:

Logistic Regression:

```python
from sklearn.linear_model import LogisticRegression

nsamples, nx, ny = X_train_em.shape
X_train_em_re = X_train_em.reshape((nsamples,nx*ny))

nsamples, nx, ny = X_test_em.shape
X_test_em_re = X_test_em.reshape((nsamples,nx*ny))

model = LogisticRegression(max_iter=1000)

model.fit(X_train_em_re, y_train_em)
```

```python
print(f"{((1 - model.score(X_train_em_re, y_train_em)) * 100):.2f}% training error")
print(f"{((1 - model.score(X_test_em_re, y_test_em)) * 100):.2f}% testing error")
```

CNN:

```python
from torchvision import datasets, transforms
from torch.optim.lr_scheduler import StepLR

class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.layer1 = torch.nn.Sequential(
            torch.nn.Conv2d(1, 32, kernel_size=3, stride=1, padding=1),
            torch.nn.ReLU(),
            torch.nn.MaxPool2d(kernel_size=2, stride=2))
        self.layer2 = torch.nn.Sequential(
            torch.nn.Conv2d(32, 64, kernel_size=3, stride=1, padding=1),
            torch.nn.ReLU(),
            torch.nn.MaxPool2d(kernel_size=2, stride=2))
        self.layer3 = torch.nn.Sequential(
            torch.nn.Conv2d(64, 128, kernel_size=3, stride=1, padding=1),
            torch.nn.ReLU(),
            torch.nn.MaxPool2d(kernel_size=2, stride=2, padding=1))
        self.fc1 = torch.nn.Linear(4 * 4 * 128, 625, bias=True)
        torch.nn.init.xavier_uniform(self.fc1.weight)
        self.layer4 = torch.nn.Sequential(
            self.fc1,
            torch.nn.ReLU())
        self.fc2 = torch.nn.Linear(625, 10, bias=True)
        torch.nn.init.xavier_uniform_(self.fc2.weight)

    def forward(self, x):
        out = self.layer1(x)
        out = self.layer2(out)
        out = self.layer3(out)
        out = out.view(out.size(0), -1)   # Flatten them for FC
        out = self.fc1(out)
        out = self.fc2(out)
        return out
    
net = Net()

optimizer = optim.Adadelta(net.parameters(), lr=1.)

criterion = nn.NLLLoss()

train_dataset = torch.utils.data.TensorDataset(torch.from_numpy(X_train_em), torch.from_numpy(y_train_em))

for inputs, target in train_dataset:
    output = net(inputs)
    loss = criterion(output, target)
    print(round(loss.item(), 2))

    net.zero_grad()
    loss.backward()
    optimizer.step()
```
