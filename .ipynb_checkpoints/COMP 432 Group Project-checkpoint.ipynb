Temporary Use of EMNIST Dataset:

```python
train_data = datasets.EMNIST(
    root = 'data',
    split = 'balanced',
    train = True,                         
    transform = ToTensor(), 
    download = True,            
)
test_data = datasets.EMNIST(
    root = 'data', 
    split = 'balanced',
    train = False, 
    transform = ToTensor()
)

loaders = {
    'train' : torch.utils.data.DataLoader(train_data, 
                                          batch_size=100, 
                                          shuffle=True, 
                                          num_workers=1),
    
    'test'  : torch.utils.data.DataLoader(test_data, 
                                          batch_size=100, 
                                          shuffle=True, 
                                          num_workers=1),
}
```

Dataset Creation: https://stackoverflow.com/questions/66711654/converting-image-into-mnist-format
                  https://github.com/gskielian/JPG-PNG-to-MNIST-NN-Format/blob/master/convert-images-to-mnist-format.py
                  https://www.youtube.com/watch?v=iwoLAj-SZ80

```python
#Get Image Directory
cwd = Path.cwd()
imgdir = os.path.join(cwd, 'img')

customImages = None
customLabels = None

for filename in os.listdir(imgdir):
    filepath = os.path.join(imgdir, filename)
    img = cv2.imread(filepath, cv2.IMREAD_GRAYSCALE)
    img_resized = cv2.resize(img, (28, 28), interpolation=cv2.INTER_LINEAR)
    img_resized = cv2.bitwise_not(img_resized)
    img_resized = np.copy(img_resized).reshape(1,28,28)
    img_final = torch.tensor(img_resized, dtype=torch.float32)
    
    img_label = torch.tensor(0)
    
    if customImages is None:
        customImages = img_final
        customLabels = img_label
    else:
        customImages = torch.stack((customImages, img_final))
        customLabels = torch.stack((customLabels, img_label))
        
print(customImages.shape)
print(customImages[0].shape)

customData = (customImages, customLabels)
```

CNN Training:

```python
def train(num_epochs, cnn, loaders):
    cnn.train()
        
    # Train the model
    total_step = len(loaders['train'])
        
    for epoch in range(num_epochs):
        for i, (images, labels) in enumerate(loaders['train']):
            
            # gives batch data, normalize x when iterate train_loader
            b_x = Variable(images)
            b_y = Variable(labels)
            output = cnn(b_x)[0]               
            loss = loss_func(output, b_y)
            
            # clear gradients for this training step   
            optimizer.zero_grad()           
            
            # backpropagation, compute gradients 
            loss.backward()           
            optimizer.step()
            
            if (i+1) % 100 == 0:
                print ('Epoch [{}/{}], Step [{}/{}], Loss: {:.4f}'
                       .format(epoch + 1, num_epochs, i + 1, total_step, loss.item()))
```

CNN Definition: https://medium.com/@nutanbhogendrasharma/pytorch-convolutional-neural-network-with-mnist-dataset-4e8a4265e118

to be modified but it works

```python
class CNN_CHARACTER(nn.Module):
    def __init__(self):
        super(CNN_CHARACTER, self).__init__()
        self.conv1 = nn.Sequential(
            nn.Conv2d(
                in_channels=1,
                out_channels=32,
                kernel_size=7,
                stride=1,
                padding=3,
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),
        )
        self.conv2 = nn.Sequential(
            nn.Conv2d(
                in_channels=32,
                out_channels=64,
                kernel_size=7,
                stride=1,
                padding=3,
            ),
            nn.ReLU(),
            nn.MaxPool2d(2),
        )
        self.out = nn.Linear(64 * 7 * 7, 47)
        
    def forward(self, x):
        x = self.conv1(x)
        x = self.conv2(x)
        x = x.view(x.size(0), -1)
        output = self.out(x)
        return output, x

cnn_c = CNN_CHARACTER()
loss_func = nn.CrossEntropyLoss()
optimizer = optim.Adam(cnn_c.parameters(), lr = 0.01)
```

Author CNN Definition

```python
class CNN_AUTHOR(nn.Module):
    def __init__(self):
        super(CNN_AUTHOR, self).__init__()
        self.conv1 = nn.Sequential(
            nn.Conv2d(
                in_channels=1,
                out_channels=32,
                kernel_size=7,
                stride=1,
                padding=3,
            ),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2),
        )
        self.conv2 = nn.Sequential(
            nn.Conv2d(
                in_channels=32,
                out_channels=64,
                kernel_size=7,
                stride=1,
                padding=3,
            ),
            nn.ReLU(),
            nn.MaxPool2d(2),
        )
        self.out = nn.Linear(64 * 7 * 7, 2)
        
    def forward(self, x):
        x = self.conv1(x)
        x = self.conv2(x)
        x = x.view(x.size(0), -1)
        output = self.out(x)
        return output, x

cnn_a = CNN_AUTHOR()
loss_func = nn.CrossEntropyLoss()
optimizer = optim.Adam(cnn_a.parameters(), lr = 0.01)
```
